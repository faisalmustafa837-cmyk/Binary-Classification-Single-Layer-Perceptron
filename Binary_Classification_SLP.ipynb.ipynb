```python


import numpy as np
import pandas as pd


# 1. LOAD DATASET


url = "https://archive.ics.uci.edu/ml/machine-learning-databases/00267/data_banknote_authentication.txt"

columns = [
    "Variance",
    "Skewness",
    "Curtosis",
    "Entropy",
    "Class"
]

df = pd.read_csv(url, header=None, names=columns)

print("Dataset Loaded Successfully!")
print("\nFirst Five Rows:")
print(df.head())

print("\nDataset Shape:", df.shape)


# 2. DATA PREPROCESSING


# Separate input features and target
X = df.iloc[:, :-1].values
y = df.iloc[:, -1].values.astype(int)

# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())


# 3. TRAIN-TEST SPLIT (80% TRAINING, 20% TESTING)


np.random.seed(42)

# Stratified splitting to maintain class proportions
train_indices = []
test_indices = []

for class_label in np.unique(y):

    class_indices = np.where(y == class_label)[0]

    np.random.shuffle(class_indices)

    split_point = int(len(class_indices) * 0.80)

    train_indices.extend(class_indices[:split_point])
    test_indices.extend(class_indices[split_point:])

train_indices = np.array(train_indices)
test_indices = np.array(test_indices)

np.random.shuffle(train_indices)

X_train = X[train_indices]
X_test = X[test_indices]

y_train = y[train_indices]
y_test = y[test_indices]

print("\nTraining Samples:", len(X_train))
print("Testing Samples:", len(X_test))


# 4. FEATURE SCALING (STANDARDIZATION)


# Calculate mean and standard deviation from training data
mean = np.mean(X_train, axis=0)
std = np.std(X_train, axis=0)

# Prevent division by zero
std[std == 0] = 1

# Apply scaling
X_train = (X_train - mean) / std
X_test = (X_test - mean) / std



class SingleLayerPerceptron:

    def __init__(self, learning_rate=0.01, epochs=100):

        self.learning_rate = learning_rate
        self.epochs = epochs
        self.weights = None
        self.bias = 0

    # Step activation function
    def activation(self, z):

        return 1 if z >= 0 else 0

    # Training function
    def fit(self, X, y):

        samples, features = X.shape

        # Initialize weights and bias
        self.weights = np.zeros(features)
        self.bias = 0

        # Training loop
        for epoch in range(self.epochs):

            for i in range(samples):

                # Calculate weighted sum
                z = np.dot(X[i], self.weights) + self.bias

                # Predict output
                prediction = self.activation(z)

                # Calculate error
                error = y[i] - prediction

                # Update weights
                self.weights += (
                    self.learning_rate * error * X[i]
                )

                # Update bias
                self.bias += self.learning_rate * error

        print("\nModel Training Completed!")

    # Prediction function
    def predict(self, X):

        predictions = []

        for sample in X:

            z = np.dot(sample, self.weights) + self.bias

            prediction = self.activation(z)

            predictions.append(prediction)

        return np.array(predictions)



# 6. TRAIN THE MODEL


model = SingleLayerPerceptron(
    learning_rate=0.01,
    epochs=100
)

model.fit(X_train, y_train)

print("\nLearned Weights:")
print(model.weights)

print("\nLearned Bias:")
print(model.bias)


# 7. TEST THE MODEL


y_pred = model.predict(X_test)

print("\nActual Values (First 15):")
print(y_test[:15])

print("\nPredicted Values (First 15):")
print(y_pred[:15])


# 8. EVALUATION METRICS


# Calculate confusion matrix manually
TP = np.sum((y_test == 1) & (y_pred == 1))
TN = np.sum((y_test == 0) & (y_pred == 0))
FP = np.sum((y_test == 0) & (y_pred == 1))
FN = np.sum((y_test == 1) & (y_pred == 0))

# Accuracy
accuracy = (TP + TN) / len(y_test)

# Precision
precision = TP / (TP + FP) if (TP + FP) > 0 else 0

# Recall
recall = TP / (TP + FN) if (TP + FN) > 0 else 0

# F1 Score
f1_score = (
    2 * precision * recall / (precision + recall)
    if (precision + recall) > 0 else 0
)


# 9. DISPLAY FINAL RESULTS


print("\n======================================")
print("       FINAL MODEL RESULTS")
print("======================================")

print(f"Accuracy  : {accuracy * 100:.2f}%")
print(f"Precision : {precision * 100:.2f}%")
print(f"Recall    : {recall * 100:.2f}%")
print(f"F1 Score  : {f1_score * 100:.2f}%")

print("\nConfusion Matrix:")
print("                 Predicted")
print("                 0       1")
print(f"Actual 0        {TN}      {FP}")
print(f"Actual 1        {FN}      {TP}")

print("\n======================================")
print("       CLASSIFICATION COMPLETED")
print("======================================")
```
